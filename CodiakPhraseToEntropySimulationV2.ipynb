## Derive a phase from character frequencies
The adapter applies NFC normalization, preserves case and whitespace, and counts Unicode code points. Character order is discarded.
It computes $H=-\sum p\log_2 p$, then divides by $\log_2(n)$ where $n$ is the normalized text length. Repeating a phrase can change this normalized value even when its character probabilities stay the same.
The phase is $\pi H/\log_2(n)$, bounded to $[0,\pi]$. Empty and single-character inputs produce phase zero. The returned analysis contains no raw phrase.

```python
from entropy import analyze_phrase

analysis = analyze_phrase("Codiak")
analysis, build_expression(analysis["phase"])
```
